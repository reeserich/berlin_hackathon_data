```python
import numpy as np
import pandas as pd
import scipy
import re
import os
from pathlib import Path
```

Downloaded from https://ewl.fenqubiao.com/, exported into Excel

```python
collect = []
# 2025
current_df = pd.read_excel('../data/260513_cas_ewl/260513_cas_ewl.xlsx', sheet_name='Early Warning Journal List 2025')
current_df.columns = current_df.loc[0]
current_df = current_df.loc[1:]
current_df['edition'] = '2025'
current_df['journal_title'] = current_df['Journal Title']
current_df['p_issn'] = current_df['ISSN/EISSN']
current_df['indexing_status'] = 'indexed'
current_df['notes_on_indexing_status'] = current_df['Warning Reason']
collect.append(current_df)

# 2024
current_df = pd.read_excel('../data/260513_cas_ewl/260513_cas_ewl.xlsx', sheet_name='Early Warning Journal List 2024')
current_df.columns = current_df.loc[0]
current_df = current_df.loc[1:]
current_df['edition'] = '2024'
current_df['journal_title'] = current_df['Journal Title']
current_df['p_issn'] = current_df['ISSN/EISSN']
current_df['indexing_status'] = 'indexed'
current_df['notes_on_indexing_status'] = current_df['Warning Reason']
collect.append(current_df)


# 2023
current_df = pd.read_excel('../data/260513_cas_ewl/260513_cas_ewl.xlsx', sheet_name='Early Warning Journal List 2023')
current_df.columns = current_df.loc[0]
current_df = current_df.loc[1:]
current_df['edition'] = '2023'
current_df['journal_title'] = current_df['Journal Title']
#current_df['p_issn'] = current_df['ISSN/EISSN']
current_df['indexing_status'] = 'indexed'
current_df['notes_on_indexing_status'] = current_df['Warning Level']
collect.append(current_df)

# 2021
current_df = pd.read_excel('../data/260513_cas_ewl/260513_cas_ewl.xlsx', sheet_name='Early Warning Journal List 2021')
current_df.columns = current_df.loc[0]
current_df = current_df.loc[1:]
current_df['edition'] = '2021'
current_df['journal_title'] = current_df['Journal Title']
#current_df['p_issn'] = current_df['ISSN/EISSN']
current_df['indexing_status'] = 'indexed'
current_df['notes_on_indexing_status'] = current_df['Warning Level']
collect.append(current_df)

# 2020
current_df = pd.read_excel('../data/260513_cas_ewl/260513_cas_ewl.xlsx', sheet_name='Early Warning Journal List 2020')
current_df.columns = current_df.loc[0]
current_df = current_df.loc[1:]
current_df['edition'] = '2020'
current_df['journal_title'] = current_df['Journal Title']
#current_df['p_issn'] = current_df['ISSN/EISSN']
current_df['indexing_status'] = 'indexed'
current_df['notes_on_indexing_status'] = current_df['Warning Level']
collect.append(current_df)
```

```python
df_all = pd.concat(collect)
df_all['e_issn'] = np.nan
df_all['internal_identifier'] = np.nan
df_all['service'] = 'cas_ewl'
```

```python
df_all_slice = df_all[['service', 'edition', 'internal_identifier', 
                       'journal_title', 'p_issn', 'e_issn', 
                       'indexing_status', 'notes_on_indexing_status']].copy()
```

```python
file_prefix = '260513_cas_ewl'
df_all_slice.to_csv('../data/' + file_prefix + '.csv', index=False)
df_all_slice.to_parquet('../data/' + file_prefix + '.parquet')
```
