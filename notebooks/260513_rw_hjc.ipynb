```python
import numpy as np
import pandas as pd
import scipy
import re
import os
from pathlib import Path
```

Downloaded from https://retractionwatch.com/the-retraction-watch-hijacked-journal-checker/ on 13 May 2026

```python
df = pd.read_excel('../data/260513_rw_hjc/Retraction Watch Hijacked Journals Checker.xlsx', header=1)
```

```python
def process_issn(issn):
    issn = str(issn)
    if issn != 'nan':
        return issn.split(', ')
    else:
        return np.nan, np.nan
```

```python
df['journal_title'] = df['Hijacked Journal Title']
df['ISSN'] = df['ISSN (Hijacked)'].apply(process_issn)
df['p_issn'] = df['ISSN'].apply(lambda x: x[0])
df['e_issn'] = df['ISSN'].apply(lambda x: x[-1])
df['service'] = 'rw_hjc'
df['edition'] = '260513'
df['internal_identifier'] = df['Unnamed: 0']
df['indexing_status'] = 'indexed'
df['notes_on_indexing_status'] = 'ISSN (Original)'
```

```python
df_all_slice = df[['service', 'edition', 'internal_identifier', 
                       'journal_title', 'p_issn', 'e_issn', 
                       'indexing_status', 'notes_on_indexing_status']].copy()
```

```python
file_prefix = '260513_rw_hjc'
df_all_slice.to_csv('../data/' + file_prefix + '.csv', index=False)
df_all_slice.to_parquet('../data/' + file_prefix + '.parquet')
```
