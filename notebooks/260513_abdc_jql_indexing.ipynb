```python
collect = []

# 2025
current_df = pd.read_excel('../data/260513_abdc/ABDC-JQL-2025-v1-260326.xlsx', sheet_name='2025 JQL', header=7)
current_df['journal_title'] = current_df['Journal Title']
current_df['p_issn'] = current_df['ISSN']
current_df['e_issn'] = current_df['ISSNOnline']
current_df['indexing_status'] = 'indexed'
current_df['notes_on_indexing_status'] = current_df['2025 rating']
current_df['edition'] = '2025'
collect.append(current_df)

# 2022
current_df = pd.read_excel('../data/260513_abdc/ABDC-JQL-2025-v1-260326.xlsx', sheet_name='2022 JQL', header=8)
current_df['journal_title'] = current_df['Journal Title']
current_df['p_issn'] = current_df['ISSN']
current_df['e_issn'] = current_df['ISSN Online']
current_df['indexing_status'] = 'indexed'
current_df['notes_on_indexing_status'] = current_df['2022 rating']
current_df['edition'] = '2022'
collect.append(current_df)

# 2019
current_df = pd.read_excel('../data/260513_abdc/ABDC-JQL-2025-v1-260326.xlsx', sheet_name='2019 JQL', header=8)
current_df['journal_title'] = current_df['Journal Title']
current_df['p_issn'] = current_df['ISSN']
current_df['e_issn'] = current_df['ISSN Online']
current_df['indexing_status'] = 'indexed'
current_df['notes_on_indexing_status'] = current_df['2019 Rating']
current_df['edition'] = '2019'
collect.append(current_df)

# 2016
current_df = pd.read_excel('../data/260513_abdc/ABDC-JQL-2025-v1-260326.xlsx', sheet_name='2016 JQL', header=7)
current_df['journal_title'] = current_df['Journal Title']
current_df['p_issn'] = current_df['ISSN']
current_df['e_issn'] = current_df['ISSN Online']
current_df['indexing_status'] = 'indexed'
current_df['notes_on_indexing_status'] = current_df['2016 rating']
current_df['edition'] = '2016'
collect.append(current_df)

# 2013
current_df = pd.read_excel('../data/260513_abdc/ABDC-JQL-2025-v1-260326.xlsx', sheet_name='2013 JQL', header=0)
current_df['journal_title'] = current_df['Journal Name']
current_df['p_issn'] = current_df['ISSN']
current_df['e_issn'] = current_df['ISSN Online']
current_df['indexing_status'] = 'indexed'
current_df['notes_on_indexing_status'] = current_df['ABDC List 2013']
current_df['edition'] = '2013'
collect.append(current_df)

# 2010
current_df = pd.read_excel('../data/260513_abdc/ABDC-JQL-2025-v1-260326.xlsx', sheet_name='2010 JQL', header=0)
current_df['journal_title'] = current_df['Journal Name']
current_df['p_issn'] = current_df['ISSN']
current_df['e_issn'] = current_df['ISSN Online']
current_df['indexing_status'] = 'indexed'
current_df['notes_on_indexing_status'] = current_df['ABDC Ranking']
current_df['edition'] = '2010'
collect.append(current_df)
```

```python
df_all = pd.concat(collect)
df_all['service'] = 'abdc'
df_all['internal_identifier'] = np.nan
```

```python
df_all_slice = df_all[['service', 'edition', 'internal_identifier', 
                       'journal_title', 'p_issn', 'e_issn', 
                       'indexing_status', 'notes_on_indexing_status']].copy()
```

```python
def process_issn(issn):
    issn = str(issn)
    if issn != 'nan':
        return issn.strip('\t')
    else:
        return np.nan
```

```python
df_all_slice['p_issn'] = df_all_slice['p_issn'].apply(process_issn)
df_all_slice['e_issn'] = df_all_slice['e_issn'].apply(process_issn)
```

```python
file_prefix = '260513_abdc'
df_all_slice.to_csv('../data/' + file_prefix + '.csv', index=False)
df_all_slice.to_parquet('../data/' + file_prefix + '.parquet')
```
