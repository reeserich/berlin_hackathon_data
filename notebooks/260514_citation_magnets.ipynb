```python
import numpy as np
import pandas as pd
import scipy
import re
import os
from pathlib import Path
```

```python
df = pd.read_excel('../data/260514_citation_magnets/High citation providers.xlsx')

df = df.dropna(subset=['DOI / link'])

df['doi'] = df['DOI / link'].str.lower()

def clean_pubid(pubid):
    if str(pubid) == 'nan':
        return np.nan
    else:
        return pubid.split('.')[-1]
    
df['dim_id'] = df['Dimensions'].apply(clean_pubid)
df['source'] = 'vehicles'
df['notes'] = df['Actions']

df = df[['source', 'doi', 'notes', 'dim_id']].copy()

file_prefix = '260514_vehicles'
df.to_csv('../data/' + file_prefix + '.csv', index=False)
#df_all_slice.to_parquet('../data/' + file_prefix + '.parquet')
```

```python
df = pd.read_excel('../data/260514_citation_magnets/High citation recipients.xlsx')

df = df.dropna(subset=['DOI / link'])

df['doi'] = df['DOI / link'].str.lower()

def clean_pubid(pubid):
    if str(pubid) == 'nan':
        return np.nan
    else:
        return pubid.split('.')[-1]
    
df['dim_id'] = df['citations'].apply(clean_pubid)
df['source'] = 'magnets'
df['notes'] = df['Unnamed: 0']

df = df[['source', 'doi', 'notes', 'dim_id']].copy()

file_prefix = '260514_magnets'
df.to_csv('../data/' + file_prefix + '.csv', index=False)
#df_all_slice.to_parquet('../data/' + file_prefix + '.parquet')
```
