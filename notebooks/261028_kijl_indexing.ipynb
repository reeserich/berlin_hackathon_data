```python
current_df = pd.read_excel('../data/261028_kijl/kijl_2026.xlsx', header=5)
```

```python
current_df['journal_title'] = current_df['Journal title']
current_df['p_issn'] = current_df['ISSN (print)']
current_df['e_issn'] = current_df['ISSN (digital)']
current_df['indexing_status'] = 'indexed'
current_df['notes_on_indexing_status'] = current_df['Level']
current_df.loc[current_df['notes_on_indexing_status'] == 0, 'indexing_status'] = 'deindexed'
current_df['edition'] = '2026'
current_df['service'] = 'kijl'
current_df['internal_identifier'] = np.nan
df_all = current_df.copy()
```

```python
df_all_slice = df_all[['service', 'edition', 'internal_identifier', 
                       'journal_title', 'p_issn', 'e_issn', 
                       'indexing_status', 'notes_on_indexing_status']].copy()
```

```python
file_prefix = '261028_kijl'
df_all_slice.to_csv('../data/' + file_prefix + '.csv', index=False)
df_all_slice.to_parquet('../data/' + file_prefix + '.parquet')
```
