```python
import numpy as np
import pandas as pd
import scipy
import re
import os
from pathlib import Path
```

Downloaded from https://jfp.csc.fi/jufoportal on 28 October 2026

```python
df = pd.read_csv('../data/261028_jufo/jfp-export.csv', encoding='utf-8')
```

```python
df = df[df['Type_en'] != 'Book publisher'].copy()
df = df[df['Level'].astype(str) != 'nan'].copy()
```

```python
df['Level'].value_counts()
```

```python
df['journal_title'] = df['Name']
df['internal_identifier'] = df['Jufo_ID']
df['service'] = 'jufo'
df['edition'] = '261028'
df['p_issn'] = df['ISSNL']
df['e_issn'] = df['ISSN2']
df['indexing_status'] = 'indexed'
df['notes_on_indexing_status'] = df['Level'].astype(str)
df.loc[df['notes_on_indexing_status'] == 'Other identified publication channels', 'indexing_status'] = 'deindexed' 
```

```python
df_all_slice = df[['service', 'edition', 'internal_identifier', 
                       'journal_title', 'p_issn', 'e_issn', 
                       'indexing_status', 'notes_on_indexing_status']].copy()
```

```python
file_prefix = '261028_jufo'
df_all_slice.to_csv('../data/' + file_prefix + '.csv', index=False)
df_all_slice.to_parquet('../data/' + file_prefix + '.parquet')
```
