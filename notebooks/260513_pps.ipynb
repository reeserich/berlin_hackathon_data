```python
import numpy as np
import pandas as pd
import scipy
import re
import os
from pathlib import Path
```

```python
os.listdir('../data/260513_pps/')
```

```python
collect = []
for filename, notes_col in zip(['annulled.csv',
 'citejacked.csv',
 'concerning.csv',
 'feet_of_clay.csv',
 'mathgen.csv',
 'problematic_cell_lines.csv',
 'scigen.csv',
 'seek_blastn.csv',
 'suspect.csv',
 'tortured.csv'], ['Decision', 'Citejacked Fingerprints', 
                   'Decision', 'Cited Problematic Papers', 
                   'Mathgen Fingerprints', 'Concern', 
                   'SCIgen Fingerprints', 
                   'Corpus from Published Papers', 
                   'Suspect Fingerprints',
                   'Tortured Phrases']):
    print(filename)
    f = pd.read_csv('../data/260513_pps/' + filename)
    print(f.columns)
    f['notes'] = f[notes_col]
    source = filename.split('.')[0]
    f['source'] = source
    collect.append(f)
df = pd.concat(collect)
```

```python
df['doi'] = df['DOI'].str.lower()
```

```python
df['dim_id'] = df['Pubid'].apply(lambda x: x.split('.')[-1])
```

```python
df = df[['source', 'doi', 'notes', 'dim_id']].copy()
```

```python
file_prefix = '260513_pps'
df.to_csv('../data/' + file_prefix + '.csv', index=False)
#df_all_slice.to_parquet('../data/' + file_prefix + '.parquet')
```
