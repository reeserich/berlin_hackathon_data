```python
rwdb = pd.read_csv('../data/260513_rwdb/retraction_watch.csv')
```

```python
rwdb = rwdb[rwdb['RetractionNature'] == 'Retraction'].copy()
```

```python
rwdb['doi'] = rwdb['OriginalPaperDOI'].astype(str).str.lower()
rwdb['notes'] = rwdb['Reason']
```

```python
df_all_slice = rwdb[['doi', 'notes']].copy()
```

```python
df_all_slice['source'] = 'rwdb'
```

```python
file_prefix = '260513_rwdb'
df_all_slice.to_csv('../data/' + file_prefix + '.csv', index=False)
#df_all_slice.to_parquet('../data/' + file_prefix + '.parquet')
```
